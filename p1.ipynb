# Problem Statement
Analyze retail sales data to identify top products, revenue trends, and customer segments. Build an interactive dashboard for management to make business decisions.

```python
import pandas as pd
```

```python
df = pd.read_csv("Superstore.csv", encoding='latin1')
```

```python
df.isnull().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df = df.drop_duplicates()
```

```python
df['Profit'].fillna(0, inplace=True)
df['Region'].fillna('Unknown', inplace=True)
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')
```

```python

customer_sales = df.groupby('Customer ID')['Sales'].sum().reset_index()
customer_sales['Segment'] = pd.qcut(customer_sales['Sales'], 4, labels=['Low','Medium','High','VIP'])
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum().reset_index()
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
```

```python
region_sales = df.groupby('Region')['Sales'].sum().reset_index()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.heatmap(region_sales.set_index('Region'), annot=True, cmap='YlGnBu')
plt.title("Region-wise Sales")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_products.values, y=top_products.index)
plt.title("Top 10 Products by Sales")
plt.show()
```
